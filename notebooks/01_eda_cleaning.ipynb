The rating distribution shows how many reviews have 1, 2, 3, 4, or 5 stars.

```python
## 8. Normalizing review text

Some reviews may look identical to a person but differ because of:
- extra spaces
- line breaks
- tabs

We will normalize whitespace before checking duplicates again.
```

```python
# Create a cleaned version of the review text.
# .str converts the column into Pandas string operations.
# .replace() changes repeated whitespace into a single space.
# .strip() removes spaces at the beginning and end.

df_sentiment["reviews.text"] = (
    df_sentiment["reviews.text"]
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

```python
# Count duplicated review texts after whitespace normalization.
duplicate_count = df_sentiment["reviews.text"].duplicated().sum()

duplicate_count
```

```python
# Show all duplicated review texts after normalization.
duplicate_reviews = df_sentiment[
    df_sentiment["reviews.text"].duplicated(keep=False)
]

duplicate_reviews[
    ["name", "reviews.rating", "reviews.text"]
]
```

```python
# Sort duplicated reviews alphabetically by their text.
# This makes matching duplicate rows easier to compare.

duplicate_reviews_sorted = duplicate_reviews.sort_values("reviews.text")

# Display the text and rating together.
duplicate_reviews_sorted[
    ["reviews.text", "reviews.rating", "name"]
]
```

```python
# Group rows that have exactly the same review text.
duplicate_rating_check = (
    duplicate_reviews
    .groupby("reviews.text")["reviews.rating"]
    .nunique()
)

# Show only review texts that have more than one different rating.
conflicting_duplicates = duplicate_rating_check[
    duplicate_rating_check > 1
]

conflicting_duplicates
```

```python
# Remove duplicated review texts.
# keep="first" means we keep the first occurrence
# and remove later repetitions.

df_sentiment = df_sentiment.drop_duplicates(
    subset=["reviews.text"],
    keep="first"
)

# Check the final shape.
df_sentiment.shape
```

```python
# Confirm there are no duplicated review texts left.
df_sentiment["reviews.text"].duplicated().sum()
```

## 9. Creating sentiment labels

### Question

How can we convert the star rating into the category that we actually want the model to predict?

### Rule

- 1–2 stars → Negative
- 3 stars → Neutral
- 4–5 stars → Positive

```python
# Create a function that receives one star rating
# and returns the corresponding sentiment label.
def get_sentiment(rating):

    # Ratings 1 and 2 represent negative reviews.
    if rating == 1 or rating == 2:
        return "negative"

    # Rating 3 represents a neutral review.
    elif rating == 3:
        return "neutral"

    # Ratings 4 and 5 represent positive reviews.
    else:
        return "positive"
    
    # Test the function manually before applying it to the dataset.
print(get_sentiment(1))
print(get_sentiment(3))
print(get_sentiment(5))
   
```

```python
sentiment_labels = df_sentiment["reviews.rating"].apply(get_sentiment)
```

```python
# Save the generated sentiment labels in a new column.
df_sentiment["sentiment"] = sentiment_labels
```

```python
# Show the rating and the new sentiment label side by side.
df_sentiment[
    ["reviews.rating", "sentiment"]
].head(10)
```

```python
# Select the sentiment column.
sentiments = df_sentiment["sentiment"]

# Count how many examples exist in each sentiment class.
sentiment_counts = sentiments.value_counts()

# Display the results
sentiment_counts
```

```python
# Import the function used to divide data into training and test sets.
from sklearn.model_selection import train_test_split
```

```python
# Split the data into training and test sets.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    Y,
    test_size=0.20,
    random_state=42,
    stratify=Y
)
```

## 13. Converting text into numbers with TF-IDF

### Question

How can a machine-learning model work with text?

Models such as Logistic Regression cannot directly understand words.

TF-IDF converts review text into numerical features that a model can process.

TF-IDF gives more importance to words that are useful for distinguishing documents and less importance to words that appear everywhere.

For the training set, TF-IDF will:
1. learn the vocabulary from `X_train`
2. calculate the TF-IDF weights
3. convert each review into numerical features

For the test set, we will only transform the text using what was learned from the training set.


X_train / X_test
      ↓
TF-IDF
      ↓
X_train_tfidf / X_test_tfidf
      ↓
Logistic Regression
      ↓
predictions
      ↓
accuracy
precision
recall
F1-score
confusion matrix

MVP with 3 baseline models, after Logistic Regression we will do something like:
Model 1: Logistic Regression
Model 2: Linear SVM
Model 3: Naive Bayes

All three will use the same train/test split so the comparison is fair.

```python
# 1. Convert text into numbers with TF-IDF

# Import the TF-IDF vectorizer from scikit-learn.
from sklearn.feature_extraction.text import TfidfVectorizer

```

# sklearn → the machine learning library
# feature_extraction → tools for creating useful model inputs
# text → specifically text-related tools
# TfidfVectorizer → the tool we want

```python
# Create the TF-IDF vectorizer using the same settings as Diogo.
tfidf = TfidfVectorizer(
    max_features=10000,
    ngram_range=(1, 2)
)
```

```python
# Learn the vocabulary from the training reviews
# and convert the training text into numerical features.
X_train_tfidf = tfidf.fit_transform(X_train)
```

fit
→ learn from the training text

transform
→ convert the text into numbers

```python
# Convert the test reviews into numerical features
# using only the vocabulary learned from the training set.
X_test_tfidf = tfidf.transform(X_test)
```

```python
# Check the size of the transformed training data.
X_train_tfidf.shape
```

```python
# Check the size of the transformed test data.
X_test_tfidf.shape
```

## 14. Baseline Model 1 — Logistic Regression

### Why are we using Logistic Regression?

Logistic Regression is a classification algorithm.

In this project, we want to classify each review as:
- Negative
- Neutral
- Positive

We are starting with Logistic Regression because it is:

- fast
- commonly used with TF-IDF text features

A baseline is our first reference model. Later, we will compare its results with SVM and Naive Bayes.

### Question

Can Logistic Regression learn useful sentiment patterns from the TF-IDF features?


```python
# Import the Logistic Regression classifier.
from sklearn.linear_model import LogisticRegression
```

```python
# Create the Logistic Regression model.
logistic_model = LogisticRegression(
    max_iter=1000
)
```

```python
# Train the model using the TF-IDF training features
# and the correct sentiment labels.
logistic_model.fit(
    X_train_tfidf,
    y_train
)
```

X_train_tfidf
= numerical information from the review text

y_train
= the correct sentiment answers

The most important parts are:
- classes_ = ['negative', 'neutral', 'positive']
  The model knows there are 3 sentiment classes.
- n_features_in_ = 12589
  It received the 12,589 TF-IDF features we created earlier.
- n_iter_ = [44]
  It converged in 44 iterations, so max_iter=1000 was more than enough.
- coef_
  These are the learned weights for the TF-IDF features. You do not need to interpret all of those now.

```python
# Predict the sentiment of the unseen test reviews.
logistic_predictions = logistic_model.predict(
    X_test_tfidf
)
```

```python
# Show the first 10 predictions made by the model.
logistic_predictions[:10]
```

```python
# Show the first 10 correct sentiment labels.
y_test.iloc[:10]
```

```python
## 15. Evaluating Logistic Regression

### Question

How well does the model perform on unseen reviews?

We will evaluate it using:

- Accuracy
- Precision
- Recall
- F1-score
- Confusion Matrix

Because the dataset is highly imbalanced, accuracy alone may be misleading.
```

```python
# Import evaluation metrics from scikit-learn.
from sklearn.metrics import(
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
# Calculate the proportion of correct predictions.
logistic_accuracy = accuracy_score(
    y_test,
    logistic_predictions
)

# Display the accuracy.
logistic_accuracy
```

```python
# Show precision, recall, and F1-score
# for Negative, Neutral, and Positive classes.
logistic_report = classification_report(
    y_test,
    logistic_predictions
)

print(logistic_report)
```

```python
# Create a confusion matrix.
logistic_confusion_matrix = confusion_matrix(
    y_test,
    logistic_predictions
)

logistic_confusion_matrix
```

```python
### Finding

Logistic Regression achieved approximately 93.6% accuracy.

However, this result is misleading because the dataset is highly imbalanced.

The model performs very well on the Positive class, but poorly on the minority classes:

- Negative recall: 0.15
- Neutral recall: 0.06
- Positive recall: 1.00

This suggests the model is strongly biased toward predicting Positive reviews.

Therefore, accuracy alone is not a sufficient metric for this problem.
```
