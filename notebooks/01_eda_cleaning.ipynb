```python
df_clean = df.copy()
```

```python
columns_to_drop = [
    "reviews.didPurchase",
    "reviews.id",
    "reviews.userCity",
    "reviews.userProvince"
]

df_clean = df_clean.drop(columns=columns_to_drop)
```

```python
df_clean = df_clean.dropna(
    subset=["reviews.text", "reviews.rating"]
)
```

```python
df_clean["reviews.rating"].value_counts().sort_index()
```

```python
def rating_to_sentiment(rating):
    if rating <= 2:
        return "negative"
    elif rating == 3:
        return "neutral"
    else:
        return "positive"

df_clean["sentiment"] = df_clean["reviews.rating"].apply(rating_to_sentiment)
```

```python
df_clean["sentiment"].value_counts()
```

```python
df_clean["sentiment"].value_counts(normalize=True) * 100
```

```python
df_clean["reviews.text"] = (
    df_clean["reviews.text"]
    .astype(str)
    .str.strip()
    .str.replace(r"\s+", " ", regex=True))
```

```python
(df_clean["reviews.text"].str.len() == 0).sum()
```

```python
df_clean["reviews.text"].duplicated().sum()
```

```python
print("Shape:", df_clean.shape)

print("\nMissing values:")
print(
    df_clean[
        ["reviews.text", "reviews.rating", "sentiment"]
    ].isna().sum()
)

print("\nSentiment distribution:")
print(df_clean["sentiment"].value_counts())

print("\nDuplicated review texts:")
print(df_clean["reviews.text"].duplicated().sum())
```

```python
duplicates = df_clean[
    df_clean["reviews.text"].duplicated(keep=False)
].sort_values("reviews.text")

duplicates[
    ["name", "reviews.rating", "sentiment", "reviews.text"]
]
```

```python
df_clean = df_clean.drop_duplicates(
    subset=["reviews.text"],
    keep="first"
)
```

```python
print("Shape:", df_clean.shape)
print("Duplicated texts:", df_clean["reviews.text"].duplicated().sum())
```
